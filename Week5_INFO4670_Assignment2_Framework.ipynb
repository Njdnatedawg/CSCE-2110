## Part B · Integrate the three files

### B5 · Standardize the key & integrate
Build one **row-per-student** analysis table called **`analysis`**: start from `students_dedup`, add a standardized numeric key, and merge in a per-student summary of `activity` (e.g., total `minutes_active`).
*Hint:* make the key with `.str.replace("NU-","")` → `int`; summarize activity with `groupby(...).sum()`; then `merge`.

```python
# # Start with the cleaned student records
analysis = students_dedup.copy()

# Create a common numeric student key
analysis["student_key"] = (
    analysis["student_id"]
    .astype(str)
    .str.replace("NU-", "", regex=False)
    .astype(int)
)

# Standardize activity student IDs
activity_temp = activity.copy()

activity_temp["student_key"] = (
    activity_temp["student_id"]
    .astype(str)
    .str.replace("NU-", "", regex=False)
    .astype(int)
)

# Summarize activity for each student
activity_summary = (
    activity_temp
    .groupby("student_key", as_index=False)["minutes_active"]
    .sum()
    .rename(columns={"minutes_active": "total_minutes_active"})
)

# Merge activity totals into the student table
analysis = analysis.merge(
    activity_summary,
    on="student_key",
    how="left"
)

print("Analysis rows:", len(analysis))
print("Unique students:", analysis["student_key"].nunique())

print("Analysis shape:", analysis.shape)
analysis.head()
```

```python
# ✅ Check
try:
    assert len(analysis) == 2000 and analysis["student_id"].is_unique
    print("✅ B5 correct — one row per student, 2000 rows")
except Exception:
    print("❌ B5 not yet — analysis should have one row per student (2000)")
```
