### c) Bar plot — count of **unique transactions per item** (10 pts)

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Set the path to the selected dataset
file_path = '/content/bread_basket-2.csv'

if os.path.exists(file_path):
    bread = pd.read_csv(file_path)
else:
    raise FileNotFoundError(f"Could not find the dataset at {file_path}.")

# c) Bar plot of transaction counts per item
subtitle = "Nathan Kazadi"

item_counts = (
    bread.groupby("item")["transaction"]
    .nunique()
    .sort_values(ascending=False)
)

ax = item_counts.plot(kind="bar", figsize=(12,5))

plt.title(f"Transaction Count per Item\nSubtitle: {subtitle}")
plt.xlabel("Item")
plt.ylabel("Unique Transactions")

plt.tight_layout()
plt.show()
```
